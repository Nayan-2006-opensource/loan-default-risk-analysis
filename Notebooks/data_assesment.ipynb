Summary of the Dataset

# Dataset Overview
Dataset have 25000 rows and 22 coloumns
Each row represent one customer 
This dataset contains information about personal loan applicants and their financial profiles, credit behavior, loan details, and loan default risk.


The dataset can be broadly divided into the following categories:

### 1. Applicant Information

This includes demographic and personal details such as age, gender, marital status, state, and city tier.

### 2. Employment and Banking Information

This section contains information related to the applicant's employment type, monthly income, and banking history.

### 3. Financial and Credit Information

These variables provide information about existing loans, monthly EMI obligations, credit utilization, credit inquiries, late payments, and CIBIL score.

### 4. Loan Information

This includes details about the requested loan, such as loan amount, tenure, purpose, collateral, and interest rate.

### 5. Default Risk Information

The dataset also includes variables related to the applicant's estimated default risk and the final default outcome.

## Overall Objective

The dataset can be used to analyze the relationship between applicant characteristics, financial behavior, credit history, and the risk of loan default.


# Column wise Description

#1.> applicant_id-> this column represents the unique customer id of each customer

#2.> age-> contains age of the customer

#3.> gender-> contains gender of the person

#4.> marital_status-> contains marital status of the customer whether he or she is married/single/divorced

#5.> city_tier-> shows the category of city where the customer lives

#6.> state-> shows in which state of the country the customer lives

#7.> employment_type-> shows the employment type of the customer whether they are salaried, govt employee, private employee, business owner or self employed

#8.> bank_account_vintage_years-> shows from how many years the bank account of a particular customer is opened

#9.> monthly_income_inr-> shows the monthly income of the customer in INR

#10.> existing_loans_count-> number of active loans per customer

#11.> existing_emi_monthly_inr-> the monthly EMI of the active loans in INR per customer

#12.> credit_utilization_ratio_pct-> shows how much percentage of available credit is being used by an individual customer

#13.> num_credit_inquiries_last_6m-> number of times the customer's credit history was checked for loans or credit related applications in the last 6 months

#14.> late_payments_last_12m-> number of times the customer made a late payment in the last 12 months

#15.> loan_amount_requested_inr-> how much loan is requested to the bank by different customers

#16.> loan_tenure_months-> shows the duration of the loan repayment period in months

#17.> loan_purpose-> the reason why the customer requested the loan from the bank

#18.> collateral_provided-> whether the customer provided collateral as security to the bank or not

#19.> cibil_score-> the CIBIL score of each customer

#20.> interest_rate_offered_pct-> interest rate offered by the bank to the customer for the loan repayment

#21.> default_risk_score-> a numerical score that represents the estimated risk of a customer defaulting on the loan

#22.> default_flag-> shows whether the customer defaulted on the loan or not
"0"-> means no default
"1"-> means customer is defaulted by the bank
