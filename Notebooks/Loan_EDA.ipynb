#summary of univariate analysis on num_credit_inquiries_last_6m

1.> average number of credit inquiries in the last 6 months is around 1.3
2.> most customers have 0 to 2 inquiries, which is normal behavior
3.> a small group of customers (54 people) have 6 or more inquiries, 
which the boxplot flags as outliers
4.> initially it seems like high inquiries should mean higher risk, 
since applying for credit repeatedly can look desperate
5.> but when i checked the default rate of this high-inquiry group, 
it turned out to be quite low (~13%) - much lower than what i expected
6.> this tells me that high inquiries alone don't make someone risky - 
they might just be comparing loan offers or genuinely need credit 
    for multiple valid reasons (like a personal loan)
7.> so unlike cibil_score, this column doesn't seem to be a strong 
standalone predictor of default on its own
