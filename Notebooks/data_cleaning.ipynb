```python
#importing required libraries
import numpy as np
import pandas as pd
```

```python
#creating a copyqof the raw df before cleaning
Loan_df=pd.read_csv(r"D:\Data Analysis\Projects\Loan Analyser eda\data\india_personal_loan_default_risk_2026.csv")
Loan_df_copy=Loan_df.copy()
display(Loan_df_copy)
```

#Data Cleaning order
1.> Resolve completeness issue
2.> Resolve messy data issues
3.> Resolve validity issues
4.> Resolve Accuracy issues
5.> Resolve Consistency issues

```python
type(Loan_df_copy)
```

```python
#Handelling Missing values in bank_account_vintage_years Column
Loan_df_copy["bank_account_vintage_years"].describe()
#as mean is  higher than median this suggests that data is likely right skewed
Loan_df_copy["bank_account_vintage_years"]=Loan_df_copy["bank_account_vintage_years"].fillna(Loan_df_copy["bank_account_vintage_years"].median())
#filled the missing value in bank_account_vintage_years coloumn with its median
display(Loan_df_copy["bank_account_vintage_years"])
display(Loan_df_copy["bank_account_vintage_years"].isnull().sum())
#code succesfully exceuted hence no missing value is now present in bank_account_vintage_years columns
```

```python
#handelling missing values in  existing_emi_montly_inr
Loan_df_copy["existing_emi_monthly_inr"].describe()
display(Loan_df_copy[
    Loan_df_copy["existing_emi_monthly_inr"].isnull()
]["existing_loans_count"].value_counts())

#checking the median EMI for each existing loan count
#this helps us fill missing EMI values using the median EMI of customers with the same number of existing loans
df_1=Loan_df_copy[Loan_df_copy["existing_loans_count"]==1]
display(df_1["existing_emi_monthly_inr"].median())


df_2=Loan_df_copy[Loan_df_copy["existing_loans_count"]==2]
display(df_2["existing_emi_monthly_inr"].median())


df_3=Loan_df_copy[Loan_df_copy["existing_loans_count"]==3]
display(df_3["existing_emi_monthly_inr"].median())

df_4=Loan_df_copy[Loan_df_copy["existing_loans_count"]==4]
display(df_4["existing_emi_monthly_inr"].median())


df_5=Loan_df_copy[Loan_df_copy["existing_loans_count"]==5]
display(df_5["existing_emi_monthly_inr"].median())


df_6=Loan_df_copy[Loan_df_copy["existing_loans_count"]==6]
display(df_6["existing_emi_monthly_inr"].median())

Loan_df_copy = Loan_df_copy.dropna(subset=["existing_emi_monthly_inr"])
Loan_df_copy["existing_emi_monthly_inr"].isnull().sum()
#dropped the rows where existing_emi_montly_inr column has missing values
display(Loan_df_copy)
Loan_df_copy.shape 
```

```python
#handelling missing values in num_credit_inquiries_last_6m column
Loan_df_copy["num_credit_inquiries_last_6m"].describe()
#here Mean=1.33 and Median=1
Loan_df_copy["num_credit_inquiries_last_6m"]=Loan_df_copy["num_credit_inquiries_last_6m"].fillna(Loan_df_copy["num_credit_inquiries_last_6m"].median())
Loan_df_copy["num_credit_inquiries_last_6m"].isnull().sum() 
#0 missing values
#all the missing values in " in num_credit_inquiries_last_6m " column is now filled with its median

```

```python
display(Loan_df_copy.isnull().sum())  #now no missing values is present in any of the column of the df 
#hence completeness issue is resolved
```

#Handelling validity issues in the dataframe


```python
display(Loan_df_copy[Loan_df_copy["age"]<Loan_df_copy["bank_account_vintage_years"]].shape)
#there are 7 rows where age is less than account age this data is completely not valid and useless
#there is no bank in the world in which there customer age is less than thier customer account age

#we are going to drop these 7 rows as well
Loan_df_copy = Loan_df_copy[Loan_df_copy["age"] >= Loan_df_copy["bank_account_vintage_years"]]
display(Loan_df_copy[Loan_df_copy["age"] < Loan_df_copy["bank_account_vintage_years"]].shape) #(0,22)
#hence those 7 unvalid rows where account age is more than customer age is finaaly dropeed
Loan_df_copy.shape#(24493,22)
#507 rows dropped till now
```

```python
df=Loan_df_copy[Loan_df_copy["existing_loans_count"]==0]
display(df.shape)
#there are 8101 rows out of 24493 rows in which existing_loan_counts_ column value=0
display(df[df["existing_emi_monthly_inr"]>0].shape) #7918 rows are there where there is no existing loan but there is existing emi 
#which is wrong and not acceptable
display(df[df["existing_emi_monthly_inr"]==0].shape) #183 rows are there in which loan count=0 and emi coloumn value=0 which is true


#to resolve this those rows where existing loan count is 0 make exiisting emi coloums of these rows as 0
#existing_loans_count = 0
#existing_emi_monthly_inr should also be 0


Loan_df_copy.loc[
    Loan_df_copy["existing_loans_count"] == 0,
    "existing_emi_monthly_inr"
] = 0
df=Loan_df_copy[Loan_df_copy["existing_loans_count"]==0]
display(df[df["existing_emi_monthly_inr"]>0].shape) #(0,22)
#hence code succesfully exceuted 
#now there is no rows where loan count is zero but emi montly is non zero

```
