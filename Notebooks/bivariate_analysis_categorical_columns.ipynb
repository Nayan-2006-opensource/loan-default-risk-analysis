```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r"D:\Data Analysis\Projects\Loan Analyser eda\data\Cleaned_Dataset.csv")
```

```python
#employment type V/s defaultflag
table=pd.crosstab(df["employment_type"],df["default_flag"],normalize="index")*100
display(table)
#1.> out of 100 percent business owners 7.36 % are defaulted (making it the highest from all employment types)
#2.> out of all customers persons who are self employed professional are the least defaulted (5.50 % are defaulted )
#3.>  out of all customers having govt job  6.72% are defaulted 
#4.>  out of all customers having PSU job  5.56% are defaulted 
#5.>  out of all customers having private job  5.83 % are defaulted 
table.plot(kind="bar")
plt.show()

```
