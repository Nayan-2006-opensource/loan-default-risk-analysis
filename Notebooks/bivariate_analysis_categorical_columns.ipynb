```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r"D:\Data Analysis\Projects\Loan Analyser eda\data\Cleaned_Dataset.csv")
```

```python
#employment type V/s defaultflag
table=pd.crosstab(df["employment_type"],df["default_flag"],normalize="index")*100
display(table)
display(table.describe())
#1.> out of 100 percent business owners 7.36 % are defaulted (making it the highest from all employment types)
#2.> out of all customers persons who are self employed professional are the least defaulted (5.50 % are defaulted )
#3.>  out of all customers having govt job  6.72% are defaulted 
#4.>  out of all customers having PSU job  5.61% are defaulted 
#5.>  out of all customers having private job  5.83 % are defaulted 
#6.> standard deviation comes out to be 0.80 indicating that there is a varation of default flag across employment_type column
table.plot(kind="bar")
plt.show()
#it is quite reasonable for businessman to have more default rate due to thier unstable income and unforunate losses due to various reasons
```

```python
#city_tier V/s default_flag
table=pd.crosstab(df["city_tier"],df["default_flag"],normalize="index")*100
display(table)
table.plot(kind="bar")
plt.show()
display(table.describe())
```

#summary of city_tier V/S default_flag
1.> customers living in tier 2 town have the highest default rate of 6.49 %
2.> customers living in tier 1 city have the least default rate of 5.62 %
3.> customers living in tier 3 town have default rate of 6.38 %
4.> standard deviation comes out to be 0.47 indicating that there is a mild  varation of default_flag across city tier

5.> this pattern could be linked to structural factors rather than the 
people themselves - tier 1 cities typically have more formal sector 
jobs, better banking infrastructure, and more stable income 
documentation, which could make repayment more predictable


6.> this is a correlation, not a judgment on individuals - it reflects 
differences in economic infrastructure across city tiers, not the 
financial responsibility of people living there

```python
#state V/S default_flag
table=pd.crosstab(df["state"],df["default_flag"],normalize="index")*100
table_sort=table.sort_values(by=1,ascending=False)
display(table_sort)
display(table.describe())
table.plot(kind="bar")
plt.show()
```
