```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r"D:\Data Analysis\Projects\Loan Analyser eda\data\Cleaned_Dataset.csv")
```

```python
#employment type V/s defaultflag
table=pd.crosstab(df["employment_type"],df["default_flag"],normalize="index")*100
display(table)
#1.> out of 100 percent business owners 7.36 % are defaulted (making it the highest from all employment types)
#2.> out of all customers persons who are self employed professional are the least defaulted (5.50 % are defaulted )
#3.>  out of all customers having govt job  6.72% are defaulted 
#4.>  out of all customers having PSU job  5.61% are defaulted 
#5.>  out of all customers having private job  5.83 % are defaulted 
table.plot(kind="bar")
plt.show()
#it is quite reasonable for businessman to have more default rate due to thier unstable income and unforunate losses due to various reasons
```

```python
table=pd.crosstab(df["city_tier"],df["default_flag"],normalize="index")*100
display(table)
table.plot(kind="bar")
plt.show()
```

Summary of city_tier V/S default_flag
1.> customers living in tier 2 town have the highest default rate of 6.49 %
2.> customers living in tier 1 city have the least default rate of 5.62 %
3.> customers living in tier 3 town have default rate of 6.38 %



```python
#loan_purpose V/S default_flag
table=pd.crosstab(df["loan_purpose"],df["default_flag"],normalize="index")*100
display(table.sort_values(by=1,ascending=False))
display(table.describe())
table.plot(kind="bar")
plt.show()
```

#summary of loan_purpose V/s default_flag
1.> customers who opt for loan due to wedding are most defaulted having a default rate of 6.54%
2.> customers who opt for loan due to debt consalidation are having default rate of 6.49% 
3.> customers who opt for loan due to vehicle puposes are having default rate of 6.31 %

4.> customers who opt for loan due to medical emergency are least defaulted having default rate of only 5.60 %
5.> customers who opt for home renovation are having default rate of 5.94%
6.> customers who opt for loan due to education are having default rate of 6.01 %

default rate lies between 5.6% to 6.54 % 
standard deviation of loan_purpose V/s default_flag  is only 0.30 % that means these values are roughly very close to each other
there is no major variation of default flag across loan puposes

```python
#collateral_provided V/s default_flag
table=pd.crosstab(df["collateral_provided"],df["default_flag"],normalize="index")*100
table.sort_values(by=1,ascending=False)
display(table)
display(table.describe())
table.plot(kind="bar")
plt.show()
```

Summary of collateral_provided V/S default flag
1.> out of all the customers who didn't provided collateral 6.833% are defaulted
2.> out of all the customers who provided collateral only 3.58 % are defaulted
3.> the range of default rate for collateral_provided column is 3.58 to 6.83 % with mean of 5.21 % 
4.> standard deviation of colloteral provided v/s default flag comes out to be 2.29
that means this there is a major variation of default_flag  across collateral_provided column
