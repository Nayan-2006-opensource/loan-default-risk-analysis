```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r"D:\Data Analysis\Projects\Loan Analyser eda\data\Cleaned_Dataset.csv")
```

```python
#monthly income vs loan_requested inr
sns.scatterplot(data=df,x="monthly_income_inr",y="loan_amount_requested_inr")
plt.show()
display(df["monthly_income_inr"].corr(df["loan_amount_requested_inr"]))
#0.006
```

# Monthly income vs loan amount requested

# The correlation between monthly income and loan amount requested is 0.006,
# which indicates almost no linear relationship between the two variables.
# This means that higher monthly income does not necessarily correspond
# to a higher loan amount requested in this dataset.
