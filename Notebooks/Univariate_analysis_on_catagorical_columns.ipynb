```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#univariate Analysis on gender column
df=pd.read_csv(r"D:\Data Analysis\Projects\Loan Analyser eda\data\Cleaned_Dataset.csv")
display(df)
```

```python
#univariate Analysis on gender column
display(df["gender"].value_counts())
#Male-> 14616
#Female-> 9404
#other-> 473
df["gender"].value_counts().plot(kind="bar")
plt.show()
df["gender"].value_counts().plot(kind="pie",autopct="%1.1f%%")
plt.show()
```

#summary of analysis
1.> out of 24493 customers "14616" are Male,"9404" are Female and "473" are other
2.> the percentage of male is "59.7",female is "38.4",other is "1.9"
