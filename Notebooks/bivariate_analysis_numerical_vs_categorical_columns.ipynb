```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
df=pd.read_csv(r"D:\Data Analysis\Projects\Loan Analyser eda\data\Cleaned_Dataset.csv")
```

```python
#Age v/s default_flag
a=df.groupby("default_flag")["age"].agg(["mean","median","min","max"])
display(a)

sns.boxplot(data=df,x="default_flag",y="age")
plt.show()
```

> Summary of age v/s default_flag
> 1.> The avg age of the customers that are marked defaulted by the bank is 34.88 years
> 2.> the maximum age of the customers that has been defaulted by the bank is 62 years

> 4.> Median age of the customer that has beeen defaulted by the bank is 34 years

>.5> The avg age of the customers that are not marked defaulted by the bank is 35.33 years
> 6.> the maximum age of the customers that are not marked  defaulted by the bank is 65 years 
> 7.> the minimum age of the customers that has been not been marked defaulted by the bank is 21 years
> 8.> Median age of the customer that has not  been  markeddefaulted by the bank is 35 years

> overall the age distribution of both default and non default customers is quite similar
